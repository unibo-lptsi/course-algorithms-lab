# Plotting capacity expansion of dynamic arrays

Basic imports

````python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
````

Let's plot how capacity evolves with different algorithms for dynamic array capacity expansion


````python
# this function must return an array with n*2 elements, reporting the capacity level
# at each time step considering linear expansion, considering n appends followed by n removes
# NB: the capacities are collected in a Python list (itself a dynamic array, with amortized O(1) append)
# and converted once at the end: np.append in a loop would copy the whole array at each call (O(n^2) overall)
def linear_expansion(n, dgrow = 500, dshrink = 1000):
    caps = []
    c = 1 # capacity of an empty array (MIN_CAPACITY in the C code)
    # n appends: the size becomes s = 1, ..., n (same rules as darray_resize_linear)
    for s in range(1, n+1):
        if s > c:
            c = s + dgrow
        caps.append(c)
    # n removes: the size becomes s = n-1, ..., 0
    for s in range(n-1, -1, -1):
        if c - s > dshrink:
            c = s + dgrow
        caps.append(c)
    return np.array(caps)

# this function must return an array with n*2 elements, reporting the capacity level
# at each time step considering geometric expansion, considering n appends followed by n removes
def geometric_expansion(n, fgrow = 5., fshrink = 5.5):
    caps = []
    c = 1 # capacity of an empty array (MIN_CAPACITY in the C code)
    # n appends: the size becomes s = 1, ..., n (same rules as darray_resize_geometric)
    for s in range(1, n+1):
        if s > c:
            c = max(int(s * fgrow), s + 1) # as in C: truncation, but at least one free slot
        caps.append(c)
    # n removes: the size becomes s = n-1, ..., 0
    for s in range(n-1, -1, -1):
        if c > fshrink * s:
            c = max(int(s * fgrow), s + 1)
        caps.append(c)
    return np.array(caps)
````

````python
n = 10000
xs = np.arange(2*n)
# xs = np.append(xs, xs)

dgrow = 500
dshrink = 2000
fgrow = 2.0
fshrink = 3.5
fig, ax1 = plt.subplots(1, 1)
fig.set_size_inches(12,4)
ax1.axvline(x=n, linestyle='--')
ax1.set_ylabel("capacity")
ax1.set_xlabel("num of unit resize ops")
ax1.plot(xs, linear_expansion(n, dgrow, dshrink), label=f"linear(dgrow={dgrow},dshrink={dshrink})", color='red')
ax1.plot(xs, geometric_expansion(n, fgrow, fshrink), label=f"geometric(fgrow={fgrow},fshrink={fshrink})", color='blue')
ax1.legend(loc='upper left')
fig.show()
````
